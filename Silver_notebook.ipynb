### Data Reading

**Importing Libraries**

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

**Reading CSV Data**

**Trip type Data**

```python
df_trip_type = spark.read.format('csv')\
                    .option("header", "true")\
                    .option("inferSchema", "true")\
                    .load("abfss://bronze@nyctaxistoragesid.dfs.core.windows.net/trip_type")
```

```python
df_trip_type.display()
```

**tripzone**

```python
df_trip_zone = spark.read.format('csv')\
                    .option("header", "true")\
                    .option("inferSchema", "true")\
                    .load("abfss://bronze@nyctaxistoragesid.dfs.core.windows.net/trip_zone")
```

```python
df_trip_zone.display()
```

**Trip Data**

```python
df_trip = spark.read.format('parquet')\
                .option("header", "true")\
                .schema(myschema)\
                .option("recursiveFileLookup", "true")\
                .load("abfss://bronze@nyctaxistoragesid.dfs.core.windows.net/trips2023data")
```

```python
myschema = '''
    VendorID BIGINT,
    lpep_pickup_datetime TIMESTAMP,
    lpep_dropoff_datetime TIMESTAMP,
    store_and_fwd_flag STRING,
    RatecodeID BIGINT,
    PULocationID BIGINT,
    DOLocationID BIGINT,
    passenger_count BIGINT,
    trip_distance DOUBLE,
    fare_amount DOUBLE,
    extra DOUBLE,
    mta_tax DOUBLE,
    tip_amount DOUBLE,
    tolls_amount DOUBLE,
    ehail_fee DOUBLE,
    improvement_surcharge DOUBLE,
    total_amount DOUBLE,
    payment_type BIGINT,
    trip_type BIGINT,
    congestion_surcharge DOUBLE
'''
```

```python
df_trip.display()
```

### Data Transformation

**Taxi Trip Type**

```python
df_trip_type =df_trip_type.withColumnRenamed('description','trip_description')
df_trip_type.display()
```

```python
df_trip_type.write.format('parquet')\
        .mode('append')\
        .option('path','abfss://silver@nyctaxistoragesid.dfs.core.windows.net/trip_type')\
        .save()
```

**Trip Zone**

```python
df_trip_zone.display()
```

```python
df_trip_zone=df_trip_zone.withColumn('zone1',split(col('Zone'),'/')[0])\
                    .withColumn('zone2',split(col('Zone'),'/')[1])
                     
df_trip_zone.display()
```

```python
df_trip_zone.write.format('parquet')\
        .mode('append')\
        .option('path','abfss://silver@nyctaxistoragesid.dfs.core.windows.net/trip_zone')\
            .save()
```

**Trips**

```python
df_trip.display()
```

```python
df_trip= df_trip.withColumn('trip_date',to_date('lpep_pickup_datetime'))\
            .withColumn('trip_year',year('lpep_pickup_datetime'))\
                .withColumn('trip_month',month('lpep_pickup_datetime'))
df_trip.display()
```

```python
df_trip=df_trip.select("VendorID","PULocationID","DOLocationID","fare_amount","total_amount")
df_trip.display()
```

```python
df_trip.write.format('parquet')\
        .mode('append')\
        .option('path','abfss://silver@nyctaxistoragesid.dfs.core.windows.net/trip2023')\
            .save()
```
